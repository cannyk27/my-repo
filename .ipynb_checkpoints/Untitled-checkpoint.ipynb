Use these to calculate the Bohr radius,

$$a_0 = \frac{4\pi \epsilon_{0} \hbar^2}{m_e e^2} = \frac{\hbar}{m_e c \alpha} $$

```python
h = 6.62607015e-34
c = 299792458
pi = 3.14159265359
hbar = h/(2 * pi)
e = 1.602176634e-19
me = 9.10938370e-31
ep0 = 8.85418781e-12
```

```python
# Bohr Radius Calculation 

a0 = (4*pi*ep0*(hbar**2))/(me*(e**2))

print(a0)
```

```python
n = 1

E1 = -e**2 / (8 * pi * ep0 * a0 * n**2)

print(E1)
```

```python
E1_eV = E1 / e

print(E1_eV)
```

```python
energy_levels = [E1_eV]

print(energy_levels)
```

```python
for n in range(2, 6):
    
    E = (-e**2 / (8 * pi * ep0 * a0 * n**2)) / e
    energy_levels.append(E)
    
print(energy_levels)
```

```python
set1 = set(energy_levels)
print (set1)
```

```python
list1 = list(set1)
print(list1)
```

```python
list1.sort()
print(list1)
```

```python
list1.reverse()
print(list1)
```

```python
E_f = 2
E_i = 3
E_gamma = E_i - E_f

lamb = (h*c)/(E_gamma)
print(lamb)
print(lamb/e)
```

```python
E_i = E_i + 1
E_gamma = E_i - E_f
lamb = (h*c)/(E_gamma)
print(lamb)
print(lamb/e)
```
