```python
h = 6.62607015e-34
c = 299792458
pi = 3.14159265359
hbar = h/(2 * pi)
e = 1.602176634e-19
me = 9.10938370e-31
ep0 = 8.85418781e-12
import numpy as np
```

Use these to calculate the Bohr radius,

$$a_0 = \frac{4\pi \epsilon_{0} \hbar^2}{m_e e^2} = \frac{\hbar}{m_e c \alpha} $$

```python
# Bohr Radius Calculation 

a0 = (4*pi*ep0*(hbar**2))/(me*(e**2))

print(a0)
```

The energy of the $n$-th is given by

$$ E_n = - \frac{e^2}{8 \pi \varepsilon_0 a_0 n^2} = \frac{E_1}{n^2} $$
Make a variable for the $n=1$ energy level.


<img src="images/Energy_Levels.png" alt="two dice" width="500"/>

```python
n = 1

E1 = -e**2 / (8 * pi * ep0 * a0 * n**2)

print(E1)
```

Change to electron volts (eV) by deviding by $e$.

```python
E1_eV = E1 / e

print(E1_eV)
```

Make a list with this as the only element to store the energy levels.

```python
energy_levels = [E1_eV]

print(energy_levels)
```

Append this list with the next 5 values for $n$.

```python
for n in range(2, 6):
    
    E = (-e**2 / (8 * pi * ep0 * a0 * n**2)) / e
    energy_levels.append(E)
    
print(energy_levels)
```

Cast this list into a set to remove any duplicates you may have created.

```python
set1 = set(energy_levels)
print (set1)
```

Now cast back into a list.

```python
list1 = list(set1)
print(list1)
```

Sort the list.

```python
list1.sort()
print(list1)
```

This is however in the reverse order to what we want so reverse it.

```python
list1.reverse()
print(list1)
```

Now calculate the first two wavelengths (i=3,4 f=2) of photons emitted in the visible Balmer series of transitions.

$$ \lambda = hc/E_\gamma$$

where

$$E_\gamma = E_i - E_f $$

the final energy state for the Balmer series is 

$$ f=2 $$

<img src="images/Energy_Levels_Transitions.png" alt="two dice" width="500"/>

```python
E_f = 2
E_i = 3
E_gamma = E_i - E_f

lamb = (h*c)/(E_gamma)
print(lamb)
print(lamb/e)
```

We are in eV, to convert back to S.I, divide the wavelengths by $e$ and print.

```python
E_i = E_i + 1
E_gamma = E_i - E_f
lamb = (h*c)/(E_gamma)
print(lamb)
print(lamb/e)
```
